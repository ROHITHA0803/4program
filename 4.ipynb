```python
import pandas as pd
import numpy as np
data = pd.Series(np.random.rand(6))
index = [['a','a','b','b','c','c'],[1,2,1,2,1,2]]
data.index = pd.MultiIndex.from_arrays(index)
print(data)
print(data['b'])
print(data[:,1])
```

```python
df = data.unstack()
print(df)
print(df.stack())
```
